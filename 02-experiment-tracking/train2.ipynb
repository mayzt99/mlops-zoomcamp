```python
import os
import pickle
import click
import mlflow

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error

mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("random-forest-train")
```

```python
def load_pickle(filename: str):
    with open(filename, "rb") as f_in:
        return pickle.load(f_in)
```

```python
X_train, y_train = load_pickle("/workspaces/mlops-zoomcamp/02-experiment-tracking/output/train.pkl")
X_val, y_val = load_pickle("/workspaces/mlops-zoomcamp/02-experiment-tracking/output/val.pkl")

print(X_train.shape, y_train.shape, X_val.shape, y_val.shape)
```

```python
def run_train():

    mlflow.sklearn.autolog()

    X_train, y_train = load_pickle("/workspaces/mlops-zoomcamp/02-experiment-tracking/output/train.pkl")
    X_val, y_val = load_pickle("/workspaces/mlops-zoomcamp/02-experiment-tracking/output/val.pkl")

    #print(len(X_train), len(y_train), len(X_val), len(y_val))

    with mlflow.start_run():

        rf = RandomForestRegressor(max_depth=10, random_state=0)
        rf.fit(X_train, y_train)
        y_pred = rf.predict(X_val)

        rmse = root_mean_squared_error(y_val, y_pred)
```

```python
run_train()
```
